```python
import math

# Gradient Descent using numerical differentiation
def gradient_descent(f, x, learning_rate, iterations):
    h = 0.00001
    for i in range(iterations):
        gradient = (f(x + h) - f(x - h)) / (2 * h)
        x = x - learning_rate * gradient
    return x, f(x)

# Enter a function
print("Available functions: x**2, (x+3)**2, sin(x), cos(x), exp(x)")
expression = input("Enter function in x: ")

# Convert input into a function
def f(x):
    return eval(expression, {"__builtins__": {}}, {
        "x": x, "sin": math.sin, "cos": math.cos,
        "exp": math.exp, "sqrt": math.sqrt, "log": math.log
    })

start = float(input("Enter starting point: "))
learning_rate = float(input("Enter learning rate: "))
iterations = int(input("Enter number of iterations: "))

minimum_x, minimum_y = gradient_descent(
    f, start, learning_rate, iterations
)

print("Approximate minimum x:", round(minimum_x, 6))
print("Function value:", round(minimum_y, 6))
```
